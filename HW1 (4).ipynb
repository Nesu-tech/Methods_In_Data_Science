````python
import pandas as pd 
from pandas import DataFrame 
import numpy as np 
print("All libraries loaded successfully")

#(a)
college = pd.read_csv("/workspaces/Methods_In_Data_Science/College.csv")
college

#(b)
college2 = pd.read_csv('College.csv', index_col=0)
college3 = college.rename({'Unnamed: 0': 'College'}, axis=1)
college3 = college3.set_index('College')
college = college3
college

````

````python
#(c).Data Summary 
college.describe()
````

````python
#(d)Scatterplot Matrix 
import matplotlib.pyplot as plt

scattermatrix = pd.plotting.scatter_matrix(
    college[["Top10perc", "Apps", "Enroll"]],
    figsize=(10, 8)
)
plt.show()

````

````python
#(e) Boxplot of Outstate by Private
import seaborn as sns
sns.boxplot(x="Private", y="Outstate", data=college)
title = "Boxplot of Outstate by Private"
plt.title(title)
plt.show()
````

````python
#(f)creating a new variable "Elite"
college["Elite"] = pd.cut(college["Top10perc"], bins=[0, 50, 100], labels=["No", "Yes"])
var_counts = college["Elite"].value_counts()
print(var_counts)

sns.boxplot(x="Elite", y="Outstate", data=college)
title = "Boxplot of Outstate by Elite"
plt.title(title)
plt.show()
````

````python
#(g)Histograms with different bins of few quantitative variables 
fig = plt.figure()
hist1 = fig.add_subplot(2,2,1)
hist1.hist(college["Top10perc"], bins=10)   
hist1.set_title("Histogram of Top10perc")
hist2 = fig.add_subplot(2,2,2)
hist2.hist(college["Enroll"], bins=15)
hist2.set_title("Histogram of Enroll")
hist3 = fig.add_subplot(2,2,3)
hist3.hist(college["F.Undergrad"], bins=20)
hist3.set_title("Histogram of F.Undergrad")
hist4 = fig.add_subplot(2,2,4)
hist4.hist(college["Outstate"], bins=25)
hist4.set_title("Histogram of Outstate")
plt.tight_layout()
````

````python
#Summary of the data 
numeric_cols = college.select_dtypes(include=[np.number]).columns
correlation_matrix = college[numeric_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix of Numeric Variables")
plt.show()
            
````
